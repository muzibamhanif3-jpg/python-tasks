```python
#TASK 10: CONTACT BOOK:

contacts = {}

while True:
    print("\n--- CONTACT BOOK MENU ---")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Display All Contacts")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    #option 1: add contact:
    if choice == "1":
        name = input("enter your name: ")
        if name in contacts:
            print("error: this name already exists!")
        else:
            phone = input("enter your phone number: ")
            contacts[name] = phone
            print("contact saved successfully!")

    #option 2: search contact:
    elif choice == "2":
        name = input("enter name to search: ")
        if name in contacts:
            print(f"Name: {name}, phone: {contacts[name]}")
        else:
            print("contact not found")

    #option 3: Display all contacts:
    elif choice == "3":
        print("\n--- all contacts ---")
        if not contacts:
            print("No contacts found.")
        else:
            for name, phone in contacts.items():
                print(f"name: {name} | phone: {phone}")

    #option 4: exit:
    elif choice == "4":
        print("exiting program. Goodbye!")
        break

    else:
        print("invalid choice! please select 1, 2, 3 or 4 ")
```

```python
#TASK 11: PERSONAL EXPENSE TRACKER:

budget = float(input("Enter yur total budget:"))
total_expenses = int(input("enter total numberof expenses: "))
expenses = []
for i in range(total_expenses):
  print("\nexpense", i + 1)
  expense_name = input("enter expense name: ")
  print("categories: food, transport, shopping, bills, others")
  category = input("enter expense category: ")
  expense_amount = float(input("enter expense amount: "))
  expense ={
      "name": expense_name,
      "category": category,
      "amount": expense_amount
  }
  expenses.append(expense)
  total_spending = sum(expense["amount"]for expense in expenses)

  category_spending = {}
  for expense in expenses:
    category = expense["category"]
    amount = expense["amount"]
    if category in category_spending:
      category_spending[category] += amount
    else:
      category_spending[category] = amount

highest_expense = max(expenses, key=lambda x: x["amount"])
budget = float(input("\nenter your budget: "))

def final_report():
  print("\n ======= EXPENSE REPORT ========")
  print("total spending:", total_spending)
  print("\ncategory-wise spending:")

  for category, amount in category_spending.items():
     print(category, ":", amount)

  print("\nhighest expense:")
  print("name:", highest_expense["name"])
  print("category:", highest_expense["category"])
  print("amount:", highest_expense["amount"])

  print("\nremaining budget:", budget - total_spending)

final_report()
```

```python
#TASK 12: STORE INVENTORY MANAGER:

inventory = {
    "milk": {"price": 200, "quantity": 10},
    "bread": {"price": 100, "quantity": 3}
}

while True:
    print("\n--- STORE INVENTORY MANAGER ---")
    print("1. Add Product")
    print("2. sell Product")
    print("3. restock Product")
    print("4. search product")
    print("5. Display inventory")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ")

    #option 1: add new product
    if choice == "1":
        name = input("Enter product name: ").strip().lower()
        if name in inventory:
            print("product already exists! use restock option to add quantity")
        else:
            price = float(input("enter price: "))
            quantity = int(input("enter quantity: "))
            inventory[name] = {"price": price, "quantity": quantity}
            print("added the product!")

    #option 2: sell product:
    elif choice == "2":
        name = input("Enter product name to sell: ").strip().lower()
        if name in inventory:
            qty = int(input("enter quantity to sell: "))
            if qty <= inventory[name]["quantity"]:
                inventory[name]["quantity"] -= qty
                print(f"sold {qty} units. remaining stock: {inventory[name]['quantity']}")
            else:
                print("insufficient quantity")
        else:
            print("product not found")

    #option 3: restock product:
    elif choice == "3":
        name = input("Enter product name to restock: ").strip().lower()
        if name in inventory:
            qty = int(input("enter quantity to add: "))
            inventory[name]["quantity"] += qty
            print(f"restocked! new stock: {inventory[name]['quantity']}")
        else:
            print("product not found")

    #option 4: search product:
    elif choice == "4":
        name = input("Enter product name to search: ").strip().lower()
        if name in inventory:
            p = inventory[name]
            print(f"name: {name} | price: rs. {p['price']} | quantity: {p['quantity']}")
        else:
            print("product not found")

    #option 5: display inventory & total value:
    elif choice == "5":
        if not inventory:
            print("no products in inventory")
        else:
            print("\n--- inventory list ---")
            total_value = 0
            for name, details in inventory.items():
                price = details["price"]
                quantity  = details["quantity"]
                total_value += price * quantity
                #low stock warning if quantity < 5
                status = "low stock" if quantity < 5 else "OK"
                print(f"name: {name} | price: rs. {price} | quantity: {quantity} | status: {status}")
            print(f"\ntotal inventory value: rs. {total_value}")

    #OPTION 6: EXIT:
    elif choice == "6":
        print("goodbye!")
        break
```

```python
#TASK 13: LIBRARY BOOK MANAGEMENT SYSTEM:

book = []

def add_book():
    try:
        book_id = int(input("enter book id: "))
    except ValueError:
        print("Invalid ID! Please enter an integer.")
        return
    for b in book:
        if b["id"] == book_id:
            print("book already exists")
            return
    title = input("enter book title: ").strip()
    author = input("enter book author: ").strip()
    book.append({
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    })
    print("book added successfully")

def display_books():
    if not book:
        print("No books in library.")
        return
    for b in book:
        status = "Available" if b["available"] else "Borrowed"
        print(f"ID: {b['id']} | Title: {b['title']} | Author: {b['author']} | Status: {status}")

def search_books():
    title = input("enter title: ").lower().strip()
    found = False
    for b in book:
        if title in b["title"].lower():
            status = "Available" if b["available"] else "Borrowed"
            print(f"ID: {b['id']} | Title: {b['title']} | Author: {b['author']} | Status: {status}")
            found = True
    if not found:
        print("book not found!")

def borrow_books():
    try:
        book_id = int(input("enter book id: "))
    except ValueError:
        print("Invalid ID!")
        return
    for b in book:
        if b["id"] == book_id:
            if b["available"]:
                b["available"] = False
                print("book borrowed successfully")
            else:
                print("already borrowed!")
            return
    print("book not found!")

def return_books():
    try:
        book_id = int(input("enter book id: "))
    except ValueError:
        print("Invalid ID!")
        return
    for b in book:
        if b["id"] == book_id:
            if not b["available"]:
                b["available"] = True
                print("book returned!")
            else:
                print("This book is already in the library!")
            return
    print("book not found!")

while True:
    print("\n--- LIBRARY BOOK MANAGEMENT SYSTEM ---")
    print("1. Add Book")
    print("2. Display Books")
    print("3. search Book")
    print("4. borrow Books")
    print("5. return book")
    print("6. Exit")

    choice = input("enter choice: ")

    if choice == "1":
        add_book()
    elif choice == "2":
        display_books()
    elif choice == "3":
        search_books()
    elif choice == "4":
        borrow_books()
    elif choice == "5":
        return_books()
    elif choice == "6":
        print("goodbye!")
        break
    else:
        print("invalid choice!")
```

```python
#Task 14:Resturant ordering system:

menu= {
    "burger":500,
    "pizza":1200,
    "pasta":800,
    "drink":200,
}

order = []

print("============= Resturant menu==============")
for key,value in menu.items():
    print(f"{key} : {value}")

while True:
    item_name = input("Enter your item name: ")
    if item_name in menu:
      quantity = int(input("Enter quantity of your item: "))
      price = menu [item_name]
      sub_total = price * quantity
      order.append({
          "items": item_name,
          "quantity": quantity,
          "sub_total": sub_total,
      })
    else:
      print("This item is not available")

    choice = input("you must want to add more items? (yes/no): ")
    if choice == "no":
       break

total= 0
for item_orders in order:
  total = total + item_orders["sub_total"]

tax = total * 0.13

if total >= 3000:
  discount = total * 0.1
else:
  discount = 0
  grand_total = total + tax - discount

def receipt():
  print("\n=========== Receipt ===========")

  for item_order in order:
    print(f'{item_order["items"]} x {item_order["quantity"]} = {item_order["sub_total"]}')

  print("total: ", total)
  print("tax: ", tax)
  print("discount: ", discount)
  print("grand_total: ", grand_total)

  receipt()
```

```python
#task 15 : clinic appointment manager:
patients = []

def add_patient():
  patient_id = int(input("enter your id: "))
  for patient in patients:
    if patient["id"] == patient_id:
       print("patient already exists")
       return

  patient_name = input("enter patient name: ")
  patient_age = int(input("Enter patient age: "))
  problem = input("enter patient problem: ")
  time = int(input("Enter appointment time: "))
  emergency = input("Is Emergency patient? yes/no: ")

  patients.append({
    "id": patient_id,
    "name": patient_name,
    "age": patient_age,
    "problem": problem,
    "time": time,
    "emergency": emergency,
  })
  print("Patient added successfully!")

def search_patient():
      patient_id = int(input("enter patient id: "))
      for patient in patients:
        if patient["id"] == patient_id:
          print("patient found")
          print(f'Patient name: {patient["name"]}')
          print(f'Patient age: {patient["age"]}')
          print(f'Patient problem: {patient["problem"]}')
          print(f'Patient time: {patient["time"]}')
          print(f'Patient emergency: {patient["emergency"]}')
          return
      print("patient not found")

def display_appointment():
  print("\n=========== emergency Patients ============")
  for patient in patients:
    if patient["emergency"].lower() == "yes":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

  print("\n================ normal patient===========")
  for patient in patients:
    if patient["emergency"].lower() == "no":
      print(f'patient name: {patient["name"]}')
      print(f'patient age: {patient["age"]}')
      print(f'patient problem: {patient["problem"]}')
      print(f'patient time: {patient["time"]}')

def delete_patient():
  patient_id = int(input("enter patient id to delete: "))
  for i, patient in enumerate(patients):
    if patient["id"] == patient_id:
      del patients[i]
      print("Patient deleted successfully!")
      return
  print("Patient not found")

def count_patient():
  emergency_count = 0
  normal_count = 0
  for patient in patients:
    if patient["emergency"].lower() == "yes":
      emergency_count = emergency_count + 1
    else:
      normal_count = normal_count + 1
  print(f"Total emergency patients: {emergency_count}")
  print(f"Total normal patients: {normal_count}")
  print(f"Total patients: {len(patients)}")

while True:
    print("\n--- Clinic Appointment Manager ---")
    print("1. Add patient")
    print("2. Search patient")
    print("3. Display appointments (Normal patients)")
    print("4. Delete patient")
    print("5. Count patients")
    print("6. Exit")

    choice = input("Enter your choice: ")
    if choice == '1':
      add_patient()
    elif choice == '2':
      search_patient()
    elif choice == '3':
      display_appointment()
    elif choice == '4':
      delete_patient()
    elif choice == '5':
      count_patient()
    elif choice == '6':
      break
    else:
      print("Invalid choice")
```
